# MNIST image download

```python
from torchvision.datasets import MNIST
from PIL import Image
import io

# 1. MNIST 다운로드 및 로드
dataset = MNIST(root="./data", train=True, download=True)

for idx in range(5):
    img, label = dataset[idx]

    # PNG 바이트로 변환
    buffer = io.BytesIO()
    img.save(buffer, format="PNG")
    img_bytes = buffer.getvalue()

    print(f"idx={idx}, label={label}, size={img.size}, bytes={len(img_bytes)}")

    # 파일로 저장
    with open(f"data\MNIST\label_{label}.png", "wb") as f:
        f.write(img_bytes)
```
